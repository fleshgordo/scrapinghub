### Set up ScrapeGraphAI API Key

To use ScrapeGraphAI, you need an API key. Please generate one from the [ScrapeGraphAI website](https://scrapegraph.ai/api) and store it securely in Colab's secrets manager (accessible via the '🔑' icon in the left panel). Name the secret `SGAI_API_KEY`.

```python
# Load the API key from Colab secrets
from google.colab import userdata

SGAI_API_KEY = userdata.get('SGAI_API_KEY')
print(SGAI_API_KEY)
if not SGAI_API_KEY:
    raise ValueError("SGAI_API_KEY not found in Colab secrets. Please add it.")
```

```python
from scrapegraph_py import (
    ScrapeGraphAI, FetchConfig,
    MarkdownFormatConfig, JsonFormatConfig,
)

sgai = ScrapeGraphAI(api_key=SGAI_API_KEY)

res = sgai.scrape(
    "https://www.hydrodaten.admin.ch/de/seen-und-fluesse/stationen-und-daten/2152",
    formats=[
        MarkdownFormatConfig(mode="reader"),
        JsonFormatConfig(prompt="extract temperatur and date/time from \"letzter messwert\""),
    ],
    fetch_config=FetchConfig(
        mode="js",
        stealth=True,
        timeout=30000,
        wait=2000,
    ),
)

if res.status == "success":
    markdown = res.data.results["markdown"]["data"]
```

```python
print(markdown)
```
