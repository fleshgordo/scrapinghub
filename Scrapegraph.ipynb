### Extracting Stories with ScrapeGraphAI

This section initializes the `ScrapeGraphAI` agent using your provided API key, then proceeds to fetch and extract stories from the specified URL. Finally, it pretty prints the extracted results for easy readability.

```python
from scrapegraph_py import ScrapeGraphAI # Import the ScrapeGraphAI library

# Initialize the ScrapeGraphAI agent with the securely loaded API key
sgai = ScrapeGraphAI(api_key=SGAI_API_KEY)

# Call the extract method to fetch and process data from the specified URL
# The first argument is the natural language query for extraction, the url is the target, and mode specifies the extraction method.
result = sgai.extract(
    "extract the stories from switzerland\n",
    url="https://www.maerchenstiftung.ch/maerchen-aus-aller-welt/schweizer-maerchen-zum-lesen-und-vorlesen/maerchensuche?kategorie=sage",
    mode="normal",
)

# Check the status of the extraction and print either the data or any errors
if result.status == "success":
    print(result.data)
else:
    print(result.error)
```

```python
# Print the raw result object from the ScrapeGraphAI extraction.
# This shows the full object structure, including status and data.
print(result)
```

```python
import json

if result.status == "success":
    print(json.dumps(result.data.json_data, indent=2))
else:
    print(result.error)
```

### Lake/River Temperature Extraction

In this example, we directly extracted the current water temperature from the `hydrodaten.admin.ch` website. The `ScrapeGraphAI` agent was given a natural language prompt to "get last temp in celsius from reading" and the specific URL. The agent successfully identified the temperature value on the page and returned it as part of the `json_data` in the result object.

To make the code more flexible, we can define the prompt and URL as separate variables before calling the `sgai.extract` method.

```python
from scrapegraph_py import ScrapeGraphAI

sgai = ScrapeGraphAI(api_key=SGAI_API_KEY)

url = "https://www.hydrodaten.admin.ch/en/seen-und-fluesse/stations/2030"
prompt = "get last temp in celsius from reading"

result = sgai.extract(
    prompt,
    url=url,
    mode="normal",
)

if result.status == "success":
    print(result.data)
else:
    print(result.error)
```

```python
print(result.data.json_data["last_temperature_celsius"])
```
